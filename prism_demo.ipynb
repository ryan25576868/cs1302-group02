## 2. Python AST Analysis & Safe Step-by-Step Execution Tracing
Analyze Python source code into structured control-flow blocks, variable insights, and bounded step-by-step execution traces without running arbitrary system commands.

```python
from python_app.analysis import analyze_python

sample_python = '''
def classify_scores(scores):
    passed = 0
    for s in scores:
        if s >= 60:
            passed += 1
            print(f"Pass: {s}")
    return passed

classify_scores([85, 42, 92])
'''

result = analyze_python(sample_python)
analysis = result["analysis"]
trace = result["trace"]

print("Language:", analysis.get("language", "python"))
print("Detected Functions:", [f["name"] for f in analysis.get("functions", [])])
print("Trace Steps Recorded:", len(trace.get("steps", [])))
print("Sample Trace Output:", trace.get("outputs", []))
```

## 3. Code Complexity & Readability Advisor (`metrics_advisor`)
Evaluate cyclomatic complexity, nesting depth, readability grade (`A/B/C`), and actionable refactoring suggestions for beginners, with Markdown report export.

```python
from python_app.metrics_advisor import evaluate_code_health, format_health_markdown

health = evaluate_code_health(analysis, trace)
print(format_health_markdown(health))
```

## 4. Multi-Language Support (`tree-sitter` Backend)
Prism also supports Java, JavaScript, TypeScript, C, and C++ under the same unified analysis pipeline.

```python
from python_app.compiled_analysis import analyze_compiled_code

sample_java = '''
public class Greeter {
    public static int sumEven(int n) {
        int total = 0;
        for (int i = 0; i <= n; i++) {
            if (i % 2 == 0) total += i;
        }
        return total;
    }
}
'''

java_res = analyze_compiled_code(sample_java, language="java")
java_health = evaluate_code_health(java_res["analysis"], java_res.get("trace"))
print("Multi-language analysis keys:", list(java_res.keys()))
print("Java Complexity Grade:", java_health["grade_label"])
```

## 5. AST-Level Structural Version Comparison (`compare_engine`)
Compare two versions of a Python module by AST symbols (added, removed, modified, unchanged functions/classes) rather than noisy line-by-line text diffs.

```python
from python_app.compare_engine import compare_code_versions

v1 = '''
def helper(x):
    return x + 1

def compute(a):
    return helper(a)
'''

v2 = '''
def helper(x):
    return x + 1

def compute(a, b=2):
    return helper(a) * b

def reset():
    return 0
'''

diff_result = compare_code_versions(v1, v2, language="python")
print("Structural Diff Summary:", diff_result.get("summary", diff_result))
```
